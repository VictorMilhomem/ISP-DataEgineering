```python
from pyspark.sql.functions import current_timestamp, lit
```

```python
file_path = "/Volumes/isp-data/raw/isp-volume/BaseDPEvolucaoMensalCisp.csv"
```

```python
df_bronze = (
    spark.read
        .option("header", True)
        .option("inferSchema", False)
        .option("encoding", "UTF-8")
        .csv(file_path, header=True, sep=";")
        .withColumn("_ingestion_timestamp", current_timestamp())
        .withColumn("_source_file", lit("BaseDPEvolucaoMensalCisp.csv"))
)

(
    df_bronze.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(
            "`isp-data`.`bronze`.`isp_monthly_timeseries_cisp`"
        )
)
```
